### (0) Import libraries 

```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

### (1) Define considered shape

```python
# Function describing the shape in 2-D
# Domain:       -1 < x < 1
#               -1 < y < 1

def isInside(x,y) :
    r1 = x*x + y*y 
    r2 = x*x + (y+1)*(y+1)
    return np.less(r1,1)*np.greater(r2,1)    # Works also for arrays
```

### (2) Generate a set of random points

```python
# Number of generations  (try to increase)

Nmc = 100

# Generate uniform distribution in (X,Y)

xy_tab = np.random.uniform(-1.,1.,size=(2,Nmc))

# Calculate acceptance weights

w_tab = isInside(xy_tab[0,:],xy_tab[1,:])

# Count accepted and rejected points

i_ok = np.sum(w_tab==1)
i_fail = np.sum(w_tab==0)

print("Total accepted: ",i_ok," rejected: ",i_fail)
                           
```

### (3) Plot results

```python
# Create result graph

plt.figure(figsize=(7, 7))

# plt.plot(xy_tab[0,w_tab],xy_tab[1,w_tab],'o',markersize=3,markeredgecolor='red',markerfacecolor='red')

plt.plot(xy_tab[0,w_tab],xy_tab[1,w_tab],'o',markersize=4,markerfacecolor='blue',markeredgecolor='blue',fillstyle='full')
plt.plot(xy_tab[0,w_tab==0],xy_tab[1,w_tab==0],'o',markersize=4,markeredgecolor='cyan',fillstyle='none')


plt.title("Monte Carlo integration",size=24)
plt.xlabel('x',size=24)
plt.ylabel('y',size=24)

# Add function

ffun = np.arange(-2./3.*np.pi,+2./3.*np.pi,0.01)
plt.plot(np.sin(ffun),np.cos(ffun),'g-',linewidth=1)
ffun2 = np.arange(-1./3.*np.pi,+1./3.*np.pi,0.01)
plt.plot(np.sin(ffun2),np.cos(ffun2)-1,'g-',linewidth=1)


plt.xlim([-1.2,1.2])
plt.ylim([-1.2,1.2])
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

fname = '05_generation_2.png'
plt.savefig(fname)

plt.show()
```

```python
# Compare results with predictions

sexp = np.pi/3. + np.sqrt(3.)/2.

pmc = np.sum(w_tab)/Nmc
emc = np.sqrt(pmc*(1-pmc)/(Nmc-1))     # Uncertainty from binomial distribution

print('Expected surface: {:.5f}'.format(sexp))
print('Generation efficiency: {:.5f} +/-{:.5f}'.format(4.*pmc,4.*emc))
```
