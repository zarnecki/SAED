# Monte Carlo integration example

Effective solid angle for muon flux measurement (P8 experiment)

Author: Aleksander Filip Żarnecki (University of Warsaw)

### (0) Import libraries 

```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

### (1) Detector setup definition

```python
# Two square scintillator counters A x A cm^2

A = 15

# Placed at the distance H cm

H = 26

# Cos(theta) range

CTHmin = 0.0
CTHmax = 1.0

# To optimize efficiency you can try to set
# CTHmin = H / np.sqrt(H*H + 2*A*A)

# Corresponding reference solid angle 

SAmc = (CTHmax - CTHmin) * 2 * np.pi
```

### (2) Event generation

```python
# Number of tracks to simulate

Nmc = 10_000_000

# Track position on lower counter

X1 = np.random.uniform(0, A, size=Nmc)
Y1 = np.random.uniform(0, A, size=Nmc)

# Track direction (uniform on sphere)

PH = np.random.uniform(0, 2*np.pi, size=Nmc)
CTH = np.random.uniform(CTHmin, CTHmax, size=Nmc)
```

### (3) Track trajectory simulation

```python
# Calculate tan(theta) from cos(theta)

STH = np.sqrt(1-CTH*CTH)
TTH = STH/CTH

# Extrapolate track to upper counter

X2 = X1 + H * TTH * np.cos(PH)
Y2 = Y1 + H * TTH * np.sin(PH)

# Check if crossing upper counter

Hit = ( X2 > 0 ) * ( X2 < A ) * ( Y2 > 0 ) * ( Y2 < A )
```

### (4) Final results

```python
Nhit = np.sum(Hit)

frac = Nhit/Nmc

unc = np.sqrt(frac*(1-frac)/Nmc)   # Binomial distribution

print(f'Fraction of accepted tracks: {frac:.5f} +/- {unc:.5f}')

sa = SAmc * frac
err = SAmc * unc

print(f'Effective solid angle: {sa:.5f} +/- {err:.5f}')
```
