### (0) Import libraries 

```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

### (1) Define log-likelihood for the outcome of the experiment

Mean and sigma from 10 normal distributed measurements.

```python
Np=10

# Assumed outcome of the experiment

x0 = 1
sigma0 = 2

# Log-likelihood for the distribution parameters

def myll(mu, sig):
    
    s2 = sig**2
    val = -Np/2/s2 * (sigma0**2 + (mu-x0)**2) - Np/2 * np.log(2*np.pi*s2)

    return val
```

### (2) Define point grid

```python
# Grid for mean

Xmin= -2
Xmax=  4
dX = 0.1

Xrange = np.arange(Xmin,Xmax,dX)

# Grid for sigma 

Smin=  0.8
Smax=  5.2
dS = 0.05

Srange = np.arange(Smin,Smax,dS)
```

```python
# 2D grid

Sgrid, Xgrid = np.meshgrid(Srange,Xrange)

# Log-likelihood calculation for the grid
    
LLval = myll(Xgrid,Sgrid)
```

### (3) Plot results

```python
# Create result graph

LLmax = np.max(LLval)

plt.figure(figsize=(9, 6),layout='constrained')

plt.contour(Sgrid,Xgrid,LLval, levels=[LLmax-4.5,LLmax-2.,LLmax-0.5,LLmax-0.002],colors=['#FF0000', '#FF0000', '#FF0000', '#000000'])

plt.title("Log-likelihood contours for PDF parameters",size=24)
plt.xlabel('$\\sigma$',size=20)
plt.ylabel('$\\mu$',size=20)

plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.grid(which='major', axis='both')

fname = '06_mlm_func.png'
plt.savefig(fname)

plt.show()
```
