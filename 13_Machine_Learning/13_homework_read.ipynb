### (0) Import libraries 

```python
import numpy as np
from matplotlib import pyplot as plt
```

### (1) Read training and test data

```python
# Number of events to read

Ntrain = 10000

Ntest = 1000

# Event size

Ndim = 4

# Train data 

infile = "http://hep.fuw.edu.pl/u/zarnecki/saed25/13_homework_signal_"+str(Ntrain)+".dat"

sigtrain = np.loadtxt(infile).reshape(Ntrain,Ndim)

infile = "http://hep.fuw.edu.pl/u/zarnecki/saed25/13_homework_background_"+str(Ntrain)+".dat"

bgtrain = np.loadtxt(infile).reshape(Ntrain,Ndim)

# Test data 

infile = "http://hep.fuw.edu.pl/u/zarnecki/saed25/13_homework_test_"+str(Ntest)+".dat"

X_test = np.loadtxt(infile).reshape(Ntest,Ndim)
```

```python
# Put train samples in one data table

X_train = np.empty(shape=(2*Ntrain,Ndim))

X_train[:Ntrain,:] = sigtrain
X_train[Ntrain:,:] = bgtrain

y_train = np.empty(2*Ntrain)
y_train[:Ntrain] = +1.0
y_train[Ntrain:] = -1.0

print("Train data defined, size: ",X_train.shape)
print("Test data defined, size: ",X_test.shape)
```

### (2) Plot train data

```python
# Create result graph

Xmin = -0.05
Xmax = 1.05

plt.subplots(1,2,figsize=(16, 7),layout='constrained')

plt.suptitle("Homework 13 train data",fontsize=22)

# X2 vs X1

plt.subplot(121)

# Training points in two categories

plt.plot(sigtrain[:,0],sigtrain[:,1],'o',color='blue',ms=2)
plt.plot(bgtrain[:,0],bgtrain[:,1],'o',color='lime',ms=2)
plt.xlim([Xmin,Xmax])
plt.ylim([Xmin,Xmax])

plt.xlabel('$x_1$',size=20)
plt.ylabel('$x_2$',size=20)

plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.grid(which='major', axis='both')


# X4 vs X3 sample

plt.subplot(122)

# Test points in two categories

plt.plot(sigtrain[:,2],sigtrain[:,3],'o',color='blue',ms=2)
plt.plot(bgtrain[:,2],bgtrain[:,3],'o',color='lime',ms=2)
plt.xlim([Xmin,Xmax])
plt.ylim([Xmin,Xmax])

plt.xlabel('$x_3$',size=20)
plt.ylabel('$x_4$',size=20)

plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.grid(which='major', axis='both')


fname = '13_homework_read.png'

plt.savefig(fname)

plt.show()
```
